# **Imbalanced datasets and redressal methodology**

```python
#class imbalance >> when one class has higher percentage
#handle:
#upsampling
#downsampling
#SMOTE
```

```python
import numpy as np
import pandas as pd

np.random.seed(1)   #for reproducibility of result

no_samples = 1000
class_0_ratio = 0.9
no_class_0 = int(no_samples * class_0_ratio)
no_class_1 = 100
```

```python
print(no_samples)
print(class_0_ratio)
print(no_class_0)
print(no_class_1)
```

```python
no_class_0, no_class_1
```

```python
class_0 = {'feature1': np.random.normal(0, 1, no_class_0),
          'feature2': np.random.normal(0, 1, no_class_0),
          'target': [0]*no_class_0}
```

```python
class_0 = pd.DataFrame(class_0)
```

```python
class_0
```

```python
class_1 = {'feature1': np.random.normal(0, 1, no_class_1),
           'feature2': np.random.normal(0, 1, no_class_1),
           'target': [1] * no_class_1
           }
```

```python
class_1 = pd.DataFrame(class_1)
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop = True)
```

```python
df
```

```python
df.target.value_counts()
```

```python
#upsampling
df_minority = df[df.target == 1]
df_majority = df[df.target == 0]
```

```python
df_minority
```

```python
df_majority
```

```python
#replace >> if replace is true, it means data point can be repeated

from sklearn.utils import resample
df_minority_upsampled = resample(df_minority, replace = True, n_samples = len(df_majority), random_state = 1)
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.dtypes
```

```python
df_minority_upsampled.info()
```

```python
df_minority_upsampled
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled.target.value_counts()
```

```python
df_majority
```

```python
df_minority
```

```python
df_majority_downsampled = resample(df_majority, replace= False, n_samples= len(df_minority), random_state = 1)
```

```python
df_downsampled = pd.concat([df_minority, df_majority_downsampled])
```

```python
df_downsampled.target.value_counts()
```

```python
df_downsampled
```

# **Data interpolation**

```python
#data interpolation >> the process of estimating unknown values from the new values

#linear interpolation
#quadratic interpolation
#cubic interpolation
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


x = np.array([1,3,5,7,9])
y = np.array([2,6,10,14,18])

plt.scatter(x, y, label = "Scatter plot")
plt.legend()
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 12])

x_new = np.linspace(x,y, 15)
y_interp = np.interp(x_new, x, y)
```

```python
y_interp
```

```python
plt.scatter(x_new, y_interp)
```

```python
#cubic interpolation

from scipy.interpolate import interp1d
f = interp1d(x, y, kind = 'cubic')
```

```python
f
```

```python
x_new = np.linspace(1,5, 10)
y_interp = f(x_new)
```

```python
y_interp
```

```python
plt.scatter(x_new, y_interp)
```

```python
#polynomial interpolation
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 4, 9, 1, 25])

p = np.polyfit(x, y, 2)
```

```python
x_new = np.linspace(1, 6, 10)
y_interp = np.polyval(p, x_new)
```

```python
plt.scatter(x, y)
```

```python
plt.scatter(x_new, y_interp)
```

```python
df = pd.DataFrame({'time':['lunch', 'breakfast', 'dinner', 'lunch', 'breakfast', 'dinner', 'lunch', 'breakfast', 'dinner'],
             'total_bill': [120, 130, 90, 125, 150, 190, 160, 180, 189]})
```

```python
df
```

```python
mean_price = df.groupby('time')['total_bill'].mean().to_dict()
```

```python
mean_price
```

```python
df['time_encoded'] = df['time'].map(mean_price)
```
