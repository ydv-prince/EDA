```python
#Due to some strings value in the column its an object column but say still you want to convert to numeric value
#use pd.to_numeric, supress the error and convert to nan
```

```python
df['packed_cell_volume']= pd.to_numeric(df['packed_cell_volume'], errors = 'coerce') #coerce means supress or ignore
```

```python
df['packed_cell_volume'].dtype
```

```python
df['packed_cell_volume'].unique() #all string character is converted to nan
```

```python
df.columns
```

```python
df['white_blood_cell_count'] = pd.to_numeric(df['white_blood_cell_count'], errors = 'coerce')
```

```python
df['red_blood_cell_count'] = pd.to_numeric(df['red_blood_cell_count'], errors = 'coerce')
```

```python
df.info()
```

```python
cat_cols = [col for col in df.columns if df[col].dtype == 'object']
```

```python
cat_cols
```

```python
num_cols = [col for col in df.columns if df[col].dtype != 'object']
```

```python
num_cols
```

```python
for col in cat_cols:
  print(f"{col}: {df[col].unique()} values\n")
```

```python
df['diabetes_mellitus'].replace(to_replace= {' yes': 'yes', '\tno': 'no', '\tyes': 'yes'}, inplace = True)
```

```python
df['diabetes_mellitus'].unique()
```

```python
df['coronary_artery_disease'].replace(to_replace= {' yes': 'yes', '\tno': 'no', '\tyes': 'yes'}, inplace = True)
```

```python
df['coronary_artery_disease'].unique()
```

```python
df['class'] = df['class'].map({'ckd': 1, 'notckd': 0, 'ckd\t': 0})
#you could have also first replaced and then map with 1 and 0
```
