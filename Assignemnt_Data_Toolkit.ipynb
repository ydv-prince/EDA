# **Data Toolkit**

1. What is NumPy, and why is it widely used in Python?
- NumPy is a core Python library for numerical computing.
- It provides fast, memory-efficient arrays and mathematical functions.
- It is widely used because it is much faster than Python lists
- Supports vectorized operations
- Forms the foundation of Pandas, Scikit-learn, TensorFlow, etc.


2. How does broadcasting work in NumPy?
- Broadcasting allows NumPy to perform operations on arrays of different shapes without copying data.
- NumPy automatically expands smaller arrays to match larger ones during operations, making calculations efficient and concise.


3. What is a Pandas DataFrame?
- A DataFrame is a 2-dimensional, labeled data structure (rows and columns), similar to an Excel sheet or SQL table.
- It allows easy data manipulation, filtering, aggregation, and analysis.

4. Explain the use of the groupby() method in Pandas
- groupby() is used to split data into groups, apply an operation (mean, sum, count), and combine the results.
It is essential for aggregation and summary statistics.

5. Why is Seaborn preferred for statistical visualizations?
- Seaborn is built on Matplotlib but simpler it automatically handles statistical relationships

- It Produces attractive, informative plots with less code
- It is ideal for exploratory data analysis (EDA).

6. Differences between NumPy arrays and Python lists
- NumPy Array	Python List
- Fixed data type	Mixed data types
- Faster computations	Slower

7. What is a heatmap, and when should it be used?

8. What does “vectorized operation” mean in NumPy?

```python
#imports

import numpy as np
import pandas as pd
```

```python
#1. How do you create a 2D NumPy array and calculate the sum of each row?

arr = np.array([[2,3,4], [4,5,6]])

sum = 0
for i in arr:
  for j in i:
    sum += j
  print('Sum of Row:',sum)
  sum = 0

#usage
print('Dimension:',arr.ndim)
```

```python
#2. Write a Pandas script to find the mean of a specific column in a DataFrame.

data = {'Name': ['Ashish','Prince','Ramanand','Anand','Aarav'],
        'Age': [23,20,15,8,6],
        'Weight': [60,50,40,30,25],
        'Height': [70,67,64,36,32]}

df = pd.DataFrame(data, index=[1,2,3,4,5])
#df['Height'].mean()
df.Height.mean()
```

```python
#3. Create a scatter plot using Matplotlib.

import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(df['Height'], df['Weight'])
plt.title('Scatter Plot of Height vs Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.show()
```

```python
#4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?

import seaborn as sns
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot= True, fmt= '.2f')
plt.show()
```

```python
#5. Generate a bar plot using Plotly.

import plotly.express as px

fig = px.bar(df, x='Name', y='Weight', title='Age of Individuals')
fig.show()
```

```python
#6. Create a DataFrame and add a new column based on an existing column.

data = {'Name': ['Ashish','Prince','Ramanand','Anand','Aarav'],
        'Age': [23,20,15,8,6],
        'Weight': [60,50,40,30,25],
        'Height': [70,67,64,36,32]}

df = pd.DataFrame(data, index=[1,2,3,4,5])

df['BMI'] = df['Weight']/(df['Height'] * 0.0254)**2
df
```

```python
#7. Write a program to perform element-wise multiplication of two NumPy arrays.

arr1 = np.array([[2,0,4],[0,2,0],[0,0,2]])
arr2 = np.array([[5,0,6],[0,5,0],[0,0,5]])

arr1 * arr2
```

```python
#8. Create a line plot with multiple lines using Matplotlib.
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(df['Height'], df['Weight'], label='Weight')
plt.plot(df['Height'], df['Age'], label='Age')
plt.title('Weight and Age vs Height')
plt.xlabel('Height')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#9. Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.

data_filter = {'Product': ['A', 'B', 'C', 'D', 'E'],
             'Price': [10, 20, 15, 25, 30],
             'Quantity': [5, 10, 7, 12, 8]}

df_filter = pd.DataFrame(data_filter)

# Filter rows where 'Price' is greater than 18
filtered_df = df_filter[df_filter['Price'] > 18]
display(filtered_df)
```
